```python
import numpy as np
import matplotlib.pyplot as plt

# Constants for Plot 1
a = 1.86 #C
b = 0.23 #Delta

# Constants for Plot 2
C_0 = 0.698
d = 0.36 #Delta
C = 3.22 #C
p_0 = C_0 * np.exp(-d * 1.4)

# Plot 1 Data
x_points1 = np.array([0.00017, 0.00033, 0.0007, 0.0014])
y_1 = np.array([2.7, 2.6, 2.3, 2])

# Plot 2 Data
x_points2 = np.array([0.00033, 0.0007, 0.0014])
y_2_points = np.array([2, 1.9, 1.9])

# x range for theoretical curves
x_line = np.logspace(-4, -2, 500)
ln_inv_x = np.log(1 / x_line)

# Curve from Plot 1
y1_curve = np.log(a * (1 / x_line)**b)

# Curve from Plot 2
y2_curve = -p_0 * np.log(p_0) - (1 - p_0) * np.log(1 - p_0) + (1 - p_0) * np.log(C * ln_inv_x**d)

# Plotting
plt.figure(figsize=(9, 6))

# Plot 1: blue curve and triangle markers
plt.plot(x_line, y1_curve, label=r'Theory ($\Delta \eta = 3.2$)', color='blue', linestyle='-')
plt.scatter(x_points1, y_1, color='blue', marker='^', s=70, label=r'$\blacktriangle$ H1. $\Delta \eta = 3.2$')

# Plot 2: green curve and circle markers
plt.plot(x_line, y2_curve, label=fr'Theory ($\Delta \eta = 1.4$), $p = {p:.2f}$', color='green', linestyle='--')
plt.scatter(x_points2, y_2_points, color='green', marker='o', s=70, label=r'$\bullet$ H1. $\Delta \eta = 1.4$')

# Axis formatting
plt.xscale('log')
plt.yticks([0, 1, 2, 3, 4])
plt.xlabel('x')
plt.ylabel(r'$\mathcal{S}_{\mathrm{hadron}}$')
plt.title(r'$\mathcal{S}_{\mathrm{hadron}}$ vs $x$ ')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)

# Combined legend
plt.legend(title=r'$5 < Q^2 < 10\ \mathrm{GeV}^2$', loc='upper right', fontsize=10, title_fontsize=11)

plt.tight_layout()
plt.show()
```
